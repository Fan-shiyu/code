# Graphic types

## Graphs for one numeric variable

### Histogram

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
data = np.random.normal(loc=0, scale=1, size=1000)  # Normal distribution

# Create histogram
plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, edgecolor='black', alpha=0.7)

# Add titles and labels
plt.title('Histogram of Normally Distributed Data', fontsize=16)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.tight_layout()
plt.show()
```

### Box plot

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate the same sample data
data = np.random.normal(loc=0, scale=1, size=1000)  # Normal distribution

# Create box plot
plt.figure(figsize=(8, 6))
plt.boxplot(data, vert=True, patch_artist=True, boxprops=dict(facecolor='#1f78b4'))

# Add titles and labels
plt.title('Box Plot of Normally Distributed Data', fontsize=16)
plt.ylabel('Value', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.tight_layout()
plt.show()
```

### Violin Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Generate the same sample data
data = np.random.normal(loc=0, scale=1, size=1000)

# Create violin plot
plt.figure(figsize=(8, 6))
sns.violinplot(data=data, inner='box', color='#1f78b4')

# Add titles and labels
plt.title('Violin Plot of Normally Distributed Data', fontsize=16)
plt.ylabel('Value', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.tight_layout()
plt.show()
```

### Density (KDE) Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Generate the same sample data
data = np.random.normal(loc=0, scale=1, size=1000)

# Create density plot
plt.figure(figsize=(10, 6))
sns.kdeplot(data, fill=True, color='#1f78b4', linewidth=2)

# Add titles and labels
plt.title('Density Plot of Normally Distributed Data', fontsize=16)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.tight_layout()
plt.show()
```

## Graphs for one categorical variable

### Bar plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Generate a categorical variable
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count frequencies using pandas
category_counts = pd.Series(cat_data).value_counts()

# Create bar plot
plt.figure(figsize=(8, 6))
plt.bar(category_counts.index, category_counts.values, color='#1f78b4', edgecolor='black')

# Add titles and labels
plt.title('Bar Plot of Fruit Categories', fontsize=16)
plt.xlabel('Fruit', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)

# Show values on top of bars
for i, val in enumerate(category_counts.values):
    plt.text(i, val + 5, str(val), ha='center', fontsize=10)

# Show plot
plt.tight_layout()
plt.show()
```

### Pie chart

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse the same categorical data
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count frequencies
category_counts = pd.Series(cat_data).value_counts()

# Create pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct='%1.1f%%',  # Show percentages
    startangle=140,
    colors=plt.cm.Pastel1.colors,
    wedgeprops={'edgecolor': 'black'}
)

# Add title
plt.title('Fruit Category Distribution (Pie Chart)', fontsize=16)

# Show plot
plt.tight_layout()
plt.show()
```

### Donut chart

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse the same categorical data
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count frequencies
category_counts = pd.Series(cat_data).value_counts()

# Create pie chart
plt.figure(figsize=(8, 8))
wedges, texts, autotexts = plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Pastel2.colors,
    wedgeprops={'edgecolor': 'black'}
)

# Draw a circle at the center to make it a donut
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)

# Add title
plt.title('Fruit Category Distribution (Donut Chart)', fontsize=16)

# Format percentage labels
for autotext in autotexts:
    autotext.set_color('black')
    autotext.set_fontsize(10)

# Show plot
plt.tight_layout()
plt.show()
```

### Radar chart

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse the same categorical data
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count category frequencies
category_counts = pd.Series(cat_data).value_counts().reindex(categories)  # Keep order consistent

# Radar chart setup
labels = category_counts.index
values = category_counts.values
num_vars = len(labels)

# Repeat first value to close the radar chart
values = np.append(values, values[0])

# Compute angle for each axis
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # Close the circle

# Create radar chart
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Draw the outline
ax.plot(angles, values, color='#1f78b4', linewidth=2)
ax.fill(angles, values, color='#1f78b4', alpha=0.25)

# Add labels
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=12)

# Set y-labels and title
ax.set_yticklabels([])
ax.set_title('Fruit Category Frequency (Radar Chart)', size=16, pad=20)

# Show plot
plt.tight_layout()
plt.show()
```

### Lollipop chart

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse the same categorical data
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count category frequencies
category_counts = pd.Series(cat_data).value_counts().reindex(categories)

# Prepare data
x = np.arange(len(category_counts))
y = category_counts.values
labels = category_counts.index

# Create lollipop chart
plt.figure(figsize=(10, 6))
plt.stem(x, y, basefmt=" ", linefmt='gray', markerfmt=' ', orientation='vertical')

# Add circular markers on top of lines
plt.plot(x, y, 'o', color='#1f78b4', markersize=10)

# Customize axes
plt.xticks(x, labels)
plt.title('Fruit Category Frequency (Lollipop Chart)', fontsize=16)
plt.xlabel('Fruit', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)

# Annotate values
for i, val in enumerate(y):
    plt.text(i, val + 5, str(val), ha='center', fontsize=10)

# Show plot
plt.tight_layout()
plt.show()
```

### Rose Chart (Polar Area Chart)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Reuse the same categorical data
categories = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 500
cat_data = np.random.choice(categories, size=n, p=[0.3, 0.2, 0.4, 0.1])

# Count frequencies and ensure consistent order
category_counts = pd.Series(cat_data).value_counts().reindex(categories)

# Prepare data
labels = category_counts.index
values = category_counts.values

# Normalize angles and prepare colors
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False)
colors = plt.cm.Set3.colors[:num_vars]

# Create polar plot
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Create bars (rose petals)
bars = ax.bar(angles, values, width=2 * np.pi / num_vars, color=colors, edgecolor='black', alpha=0.8)

# Add labels
ax.set_xticks(angles)
ax.set_xticklabels(labels, fontsize=12)

# Remove radial grid labels for clean look
ax.set_yticklabels([])
ax.set_title('Fruit Category Distribution (Rose Chart)', fontsize=16, pad=20)

# Show plot
plt.tight_layout()
plt.show()
```

## Graphs for one numeric + one categorical variables

### Box Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import warnings
# Suppress all warnings
warnings.filterwarnings("ignore")
# Set random seed for reproducibility
np.random.seed(42)

# Generate categorical variable
fruits = ['Apple', 'Banana', 'Cherry', 'Orange']
n = 200
fruit_data = np.random.choice(fruits, size=n)

# Generate numeric variable (weights with different distributions per fruit)
weights = []
for fruit in fruit_data:
    if fruit == 'Apple':
        weights.append(np.random.normal(loc=150, scale=10))  # grams
    elif fruit == 'Banana':
        weights.append(np.random.normal(loc=120, scale=15))
    elif fruit == 'Cherry':
        weights.append(np.random.normal(loc=10, scale=2))
    else:  # Date
        weights.append(np.random.normal(loc=25, scale=5))

# Create DataFrame
df = pd.DataFrame({
    'Fruit': fruit_data,
    'Weight': weights
})

# Create box plot
plt.figure(figsize=(10, 6))
sns.boxplot(x='Fruit', y='Weight', data=df, palette='Set2')

# Add labels and title
plt.title('Distribution of Fruit Weights by Type', fontsize=16)
plt.xlabel('Fruit Type', fontsize=12)
plt.ylabel('Weight (grams)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.tight_layout()
plt.show()
```

### Grouped Bar Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Re-create the dataset
np.random.seed(42)
fruits = ['Apple', 'Banana', 'Cherry', 'Orange']
suppliers = ['Supplier A', 'Supplier B']
n = 200
fruit_data = np.random.choice(fruits, size=n)
supplier_data = np.random.choice(suppliers, size=n)

weights = []
for fruit in fruit_data:
    if fruit == 'Apple':
        weights.append(np.random.normal(loc=150, scale=10))
    elif fruit == 'Banana':
        weights.append(np.random.normal(loc=120, scale=15))
    elif fruit == 'Cherry':
        weights.append(np.random.normal(loc=10, scale=2))
    else:  # Date
        weights.append(np.random.normal(loc=25, scale=5))

# Create DataFrame
df = pd.DataFrame({
    'Fruit': fruit_data,
    'Supplier': supplier_data,
    'Weight': weights
})

# Compute mean weight per fruit per supplier
grouped = df.groupby(['Fruit', 'Supplier'])['Weight'].mean().unstack()

# Custom colors for the bars
custom_colors = ['#7fc97f', '#beaed4']

# Create grouped bar plot
ax = grouped.plot(kind='bar', figsize=(10, 6), edgecolor='black', color=custom_colors)

# Add labels and title
plt.title('Average Fruit Weight by Supplier', fontsize=16)
plt.xlabel('Fruit Type', fontsize=12)
plt.ylabel('Average Weight (grams)', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Supplier')

# Show plot
plt.tight_layout()
plt.show()
```

### Stacked Bar Plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Re-create the dataset
np.random.seed(42)
fruits = ['Apple', 'Banana', 'Cherry', 'Orange']
suppliers = ['Supplier A', 'Supplier B']
n = 200
fruit_data = np.random.choice(fruits, size=n)
supplier_data = np.random.choice(suppliers, size=n)

weights = []
for fruit in fruit_data:
    if fruit == 'Apple':
        weights.append(np.random.normal(loc=150, scale=10))
    elif fruit == 'Banana':
        weights.append(np.random.normal(loc=120, scale=15))
    elif fruit == 'Cherry':
        weights.append(np.random.normal(loc=10, scale=2))
    else:  # Date
        weights.append(np.random.normal(loc=25, scale=5))

# Create DataFrame
df = pd.DataFrame({
    'Fruit': fruit_data,
    'Supplier': supplier_data,
    'Weight': weights
})

# Compute sum of weights per fruit per supplier (for stacking)
grouped = df.groupby(['Fruit', 'Supplier'])['Weight'].sum().unstack().reindex(fruits)

# Custom colors
custom_colors = ['#7fc97f', '#beaed4']

# Create stacked bar plot
ax = grouped.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6),
    color=custom_colors,
    edgecolor='black'
)

# Add labels and title
plt.title('Total Fruit Weight by Supplier (Stacked)', fontsize=16)
plt.xlabel('Fruit Type', fontsize=12)
plt.ylabel('Total Weight (grams)', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Supplier')

# Show plot
plt.tight_layout()
plt.show()
```

### KDE Plots

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Re-create the dataset
np.random.seed(42)
fruits = ['Apple', 'Banana', 'Cherry', 'Orange']
suppliers = ['Supplier A', 'Supplier B']
n = 200
fruit_data = np.random.choice(fruits, size=n)
supplier_data = np.random.choice(suppliers, size=n)

weights = []
for fruit in fruit_data:
    if fruit == 'Apple':
        weights.append(np.random.normal(loc=150, scale=10))
    elif fruit == 'Banana':
        weights.append(np.random.normal(loc=120, scale=15))
    elif fruit == 'Cherry':
        weights.append(np.random.normal(loc=10, scale=2))
    else:  # Date
        weights.append(np.random.normal(loc=25, scale=5))

# Create DataFrame
df = pd.DataFrame({
    'Fruit': fruit_data,
    'Supplier': supplier_data,
    'Weight': weights
})


plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='Weight', hue='Fruit', fill=True, common_norm=False, alpha=0.4, linewidth=2)
plt.title('KDE Plot of Fruit Weights by Type', fontsize=16)
plt.xlabel('Weight (grams)')
plt.ylabel('Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
categories = ['Alpha', 'Beta', 'Gamma', 'Delta', 'Epsilon']
values = [23, 45, 12, 37, 29]
colors = ['#66c2a5', '#fc8d62', '#8da0cb', '#e78ac3', '#a6d854']

# Calculate a little padding for the y-axis limit
y_max = max(values) + 5  # Add extra space

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

# Plot 1: Without annotation
axs[0].bar(categories, values, color=colors)
#axs[0].set_title("❌ No Annotations")
axs[0].set_xlabel("Category")
axs[0].set_ylabel("Value")
axs[0].set_ylim(0, y_max)  # Set higher y-limit

# Plot 2: With annotation
axs[1].bar(categories, values, color=colors)
for i, value in enumerate(values):
    axs[1].text(i, value + 1, str(value), ha='center', va='bottom', fontsize=10, fontweight='bold')
#axs[1].set_title("✅ With Annotations")
axs[1].set_xlabel("Category")
axs[1].set_ylim(0, y_max)  # Same y-limit for consistency

# Layout and show
#plt.suptitle("The Value of Annotations in Data Visualization", fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.90])
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample time-series data
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
visits = [120, 135, 150, 145, 160, 180, 130]

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# Plot 1: No title or subtitle
axs[0].plot(days, visits, marker='o', color='#1f77b4')
axs[0].set_xlabel("Day of Week")
axs[0].set_ylabel("Visitors")
#axs[0].set_title("❌ No Title", fontsize=12)

# Plot 2: With title and subtitle
axs[1].plot(days, visits, marker='o', color='#1f77b4')
axs[1].set_xlabel("Day of Week")
axs[1].set_ylabel("Visitors")
axs[1].set_title("Website Traffic Trends", fontsize=16, fontweight='bold', pad=20)  # Move title down a bit
axs[1].text(0.5, 1.02, "Daily visitor count over the past week", 
            transform=axs[1].transAxes, ha='center', fontsize=12)

# Adjust layout and show
#plt.suptitle("The Importance of Titles in Graphs", fontsize=18, y=1.03)
plt.tight_layout(rect=[0, 0.03, 1, 0.92])
plt.show()
```

## Overlap and transparency

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# Set seed for reproducibility
np.random.seed(0)

# Create sample data with overlapping distributions
data = pd.DataFrame({
    'A': np.random.normal(loc=0, scale=1, size=500),
    'B': np.random.normal(loc=1, scale=1, size=500),
    'C': np.random.normal(loc=0.5, scale=1, size=500)
})

# Define colors for A, B, C
colors = ['#e41a1c', '#377eb8', '#4daf4a']

# Compute KDEs
kde_A = gaussian_kde(data['A'])
kde_B = gaussian_kde(data['B'])
kde_C = gaussian_kde(data['C'])

# Define common x-axis range
x_range = np.linspace(-4, 5, 1000)

# Evaluate KDEs
density_A = kde_A(x_range)
density_B = kde_B(x_range)
density_C = kde_C(x_range)

# Create the figure
plt.figure(figsize=(12, 5))

# Plot 1: Full color fill
plt.subplot(1, 2, 1)
plt.fill_between(x_range, density_A, alpha=1.0, color=colors[0], label='A')
plt.fill_between(x_range, density_B, alpha=1.0, color=colors[1], label='B')
plt.fill_between(x_range, density_C, alpha=1.0, color=colors[2], label='C')
#plt.title('Density Plot with Full Color Fill')
plt.legend()

# Plot 2: Transparent fill
plt.subplot(1, 2, 2)
plt.fill_between(x_range, density_A, alpha=0.4, color=colors[0], label='A')
plt.fill_between(x_range, density_B, alpha=0.4, color=colors[1], label='B')
plt.fill_between(x_range, density_C, alpha=0.4, color=colors[2], label='C')
#plt.title('Density Plot with Transparency')
plt.legend()

# Final layout
plt.tight_layout()
plt.show()
```
